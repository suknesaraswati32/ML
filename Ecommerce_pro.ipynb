```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing import OneHotEncoder
from sklearn.model_selection import train_test_split
```

```python
df=pd.read_csv('shop_smart_ecommerce.csv')
```

```python
df.drop(columns=['Administrative', 'Administrative_Duration', 'Informational','Informational_Duration','OperatingSystems','Browser'])
```

```python
transformer=ColumnTransformer(transformers=[
    ('ord',OrdinalEncoder(),['Weekend']),
    ('ohe',OneHotEncoder(handle_unknown='ignore'),['Month','VisitorType'])
],remainder='passthrough')
```

```python
print("number of rows:", df.shape[0])
```

```python
X = df.drop('Revenue', axis=1)
y = df['Revenue']
```

```python
X_train, X_test, y_train, y_test = train_test_split(
X, y, test_size=0.33, random_state=42)
```

```python
X_train_transformed=transformer.fit_transform(X_train)
X_test_transformed=transformer.transform(X_test)
```

```python
from sklearn.tree import DecisionTreeClassifier
clf=DecisionTreeClassifier(max_depth=12,max_leaf_nodes=12)
```

```python
model=clf.fit(X_train_transformed,y_train)
y_pred=model.predict(X_test_transformed)
```

```python
from sklearn.metrics import f1_score
print('f1_score',f1_score(y_test, y_pred))
```
