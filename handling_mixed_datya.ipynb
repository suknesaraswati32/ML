```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv('HousePricePrediction.csv',usecols=['BldgType','LotConfig','TotalBsmtSF','SalePrice'])
```

```python
df.head()
```

```python
df['BldgType'].unique()
```

```python
#exact numerical and categorical data into a single col
df['number_numerical']=pd.to_numeric(df['number'],errors='coerce',downcast='integer')
```

```python
#exact categorical data
df['number_categorical']=np.where(df['number_numerical'].isnull(),df['number'],np.nan)
```

```python
#single row have numeric as well as categoric data
df['BldgType_num']=df['BldgType'].str.extract('(\d+)')#captures numerical part
df['BldgType_cat']=df['BldgType'].str[1:len(df['BldgType'])]# capture the first letter
```
