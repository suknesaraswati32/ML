```python
import pandas as pd 
import matplotlib.pyplot as plt
import numpy as np 
from sklearn.compose import ColumnTransformer
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder,LabelEncoder
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import f1_score
```

```python
df=pd.read_csv('shop_smart_ecommerce.csv')
df.describe()
df['Info_pro']=df['Informational']+df['ProductRelated']
df['Info_pro_due']=df['Informational_Duration']+df['ProductRelated_Duration']
df.drop(columns=[
    'Informational_Duration',
    'Informational',
    'ProductRelated_Duration',
    'ProductRelated'
], inplace=True)
df.info()
```

```python
# sns.boxplot(x='Revenue', y='ProductRelated_Duration', data=df)
# sns.scatterplot(x='ProductRelated_Duration', y='Revenue', data=df)

# df['ProductRelated_Duration'].sort_values(ascending=False).head(10)
# df['Month'] = pd.to_datetime(df['Month'])
# print(df['Month'].unique())
```

```python
month_map = {
    'Feb': 2,
    'Mar': 3,
    'May': 5,
    'June': 6,
    'Jul': 7,
    'Aug': 8,
    'Sep': 9,
    'Oct': 10,
    'Nov': 11,
    'Dec': 12
}

df['Month'] = df['Month'].map(month_map)
df['Info_pro'] = np.log1p(df['Info_pro'])
df['Info_pro_due'] = np.log1p(df['Info_pro_due'])
# sns.histplot(
#     data=df,
#     x="ProductRelated_Duration",
#     bins=20
# )
```

```python
X=df.drop('Revenue',axis=1)
y=df['Revenue']
df.info()
```

```python
X_train, X_test, y_train, y_test = train_test_split(
X, y, test_size=0.33, random_state=42
)
```

```python
trf1=ColumnTransformer([
    ('ohe',OneHotEncoder(handle_unknown='ignore',drop='first'),[11]),
    ('ord',OrdinalEncoder(),[12])
],remainder='passthrough')
```

```python
X_train_encoded=trf1.fit_transform(X_train)
X_test_encoded=trf1.transform(X_test)
# print(X_train.columns)
# print(trf1)
```

```python
le = LabelEncoder()
```

```python
y_train_encoded=le.fit_transform(y_train)
y_test_encoded=le.transform(y_test)
```

```python
alphas = [0.0, 0.0001, 0.0005, 0.001, 0.005, 0.01]
for alpha in alphas:
    model= DecisionTreeClassifier(max_depth=3, random_state=42,ccp_alpha=alpha)
    model.fit(X_train_encoded,y_train_encoded)
    y_pred=model.predict(X_test_encoded)
    print(f1_score(y_test_encoded, y_pred))
```
