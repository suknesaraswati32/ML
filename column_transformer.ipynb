```python
import numpy as np
import pandas as pd
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
```

```python
df=pd.read_csv('Amazon Sale Report.csv',usecols=[3,6,9,10,12,13,14])
```

```python
df.head()
# df.isnull().sum()
X=df.drop('Courier Status',axis=1)
y=df['Courier Status']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.33, random_state=42)
```

```python
from sklearn.compose import ColumnTransformer
```

```python
transformer=ColumnTransformer(transformers=[
    ('tnf1',SimpleImputer(strategy='most_frequent'),['currency']),
    ('tnf2',OrdinalEncoder(categories=[['XS', 'S', 'M', 'L', 'XL'`, 'XXL', '3XL','4XL','5XL','6XL','Free']]),['Size']),
    ('tnf3',OneHotEncoder(sparse_output=False,drop='first'),['Category'])
],remainder='passthrough')
```

```python
transformer.fit_transform(X_train)
```

```python
transformer.fit_transform(X_test)
```
