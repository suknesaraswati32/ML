```python
#server se link se data kaisefetch kare
import requests
from io import StringIO
import pandas as pd
url="https://raw.githubusercontent.com/cs109/2014_data/master/countries.csv"
headers={'User-Agent':'Mozilla/5.0 (Macintosh;Intel Mac OS X10.14;rv:66.0)Gecko/20100101 Firefox/66.0'}
req=requests.get(url,headers=headers)
data=StringIO(req.text)
pd.read_csv(data)
```

```python
#to handel the tsv files
pd.read_csv('sara.tsv',sep='\t',names=['sno','name','release_year','rating'])
#if we have tomention particular column as an index col
pd.read_csv('sara.tsv',index_col='enrollee_id')
#if our names col get an index 0 how to remove
pd.read_csv('sara.tsv',headerd=1)
#if we want to select particular cols from data
pd.read_csv('sara.csv',usecols=['name','age','education']
#to skip particular rows
pd.read_csv('sara.tsv',skiprows=[0,1])   
#no of rows to be select
pd.read_csv('sara.tsv',nrows=100)    
#encoding
pd.read_csv('sara.tsv',encoding='latin-1')
#if particular lines have more than other rows value
pd.read_csv('sara.tsv',error_bad_lines=False) 
#to change type of col to date 
pd.read_csv('sara.tsv',parse_dates=['date']).info()       
#to specify particular val as nan
pd.read_csv('sara.tsv',na_values=['-'])            
#loding dataset in chunks
dfs=pd.read_csv('sara.tsv',chunksize=1000)
for chunks in dfs:
    print(chunk.shape)
```
